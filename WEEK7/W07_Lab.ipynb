## Topic 4 · Parsing real text
*Walkthrough: `04 - Parsing Real Text`*

### D14. Parse one messy line, step by step

Start with `raw_line = "  Srv-01 , used=87 , total=100  "`.

1. `.strip()` the whole line.
2. Split off the hostname with `.split(",", 1)` into `hostname_part` and `rest`.
3. Clean the hostname with `.strip().lower()`.
4. Split `rest` on `","`. For each piece, split on `"="`, strip the key and the value, and convert the
   value with `float()`.
5. Put everything into one dict and print it.

**Expected output:** `{'hostname': 'srv-01', 'used': 87.0, 'total': 100.0}`

````python
# D14
"""The programe passes the string using different method  stores it in a dictionary and outputs it """
raw_line = "Srv-01 , used=87, total=100"
strip_line =raw_line.strip() #using strip 
hostname_part,rest = strip_line.split(",", 1) #spliting into hostname_part and rest  
hostname = hostname_part.strip().lower() #using strip and lower 
dict = {"hostname" : hostname} #assigning a dictionary 
for piece in rest.split(","): #spliting 
    key,value = piece.split("=") #spliting into key and value 
    dict[key.strip()]= float(value.strip())
print (dict)
````

### D15. Turn it into a function

1. Put your steps from D14 into a function `parse_line(raw_line)` that **returns** the dict.
   Give it a one-line docstring.
2. Print `parse_line("SRV-02,used=120,total=100")`.

**Expected output:** `{'hostname': 'srv-02', 'used': 120.0, 'total': 100.0}`

````python
# D15
"""The program creates a function to parse raw_line using different methods"""
def parse_line(raw_line):
 strip_line =raw_line.strip() #striping the line 
 hostname_part,rest = strip_line.split(",", 1) #striping the line using , and only by 1 
 hostname = hostname_part.strip().lower()
 result_dict = {"hostname" : hostname} #creating a dictionary 

 for piece in rest.split(","):
    key,value = piece.split("=")
    result_dict[key.strip()]= float(value.strip())
 return result_dict
print(parse_line("SRV-02 , used=87, total=100"))
````

### C1. Parse a whole file

Read `messy_lines.txt` (in this `LAB` folder) line by line. Parse each line with `parse_line` from D15,
and print the hostname with `OVER LIMIT` (100% or more), `WARNING` (90% or more) or `OK`.
Do not write the cleaning logic a second time.

**Expected output:**
```
srv-01 OK
srv-02 OVER LIMIT
srv-03 OK
srv-04 WARNING
```

````python
# C1
"""The program calculates the percentage and checks the status of each value and output the hostname and status """
with open("messy_lines.txt") as f:
   for line in f:
    if line.strip(): #usign strip to remove whitespaces 
     data = parse_line(line) #parsing a line 



     percentage = data["used"]/data["total"]*100 #calculates percentage and assigning status values
     if percentage >= 100:
      status ="OVER LIMIT"
     elif percentage >= 90:
      status ="WARNING"
     else:
      status = "OK"
    print(f"{data['hostname']} {status}")
````

### C2. When parsing breaks

Loop over the three lines below. Use `try` / `except ValueError` so that lines that do not parse print
`Could not parse:` and the line, instead of crashing.

```python
lines = ["srv-07,used=50,total=100",
         "srv-08,used=unknown,total=100",
         "srv-09,note=a=b,used=1,total=2"]
```

**Expected output:**
```
{'hostname': 'srv-07', 'used': 50.0, 'total': 100.0}

```

````python
# C2
"""The program creates a list and uses try and except to output if line can be parsed or not """
lines = ["srv-07,used=50,total=100",
         "srv-08,used=unknown,total=100",
         "srv-09,note=a=b,used=1,total=2"]
for line in lines:
    try:
        data = parse_line(line) #using try 
        print(data)
    except ValueError as e: #using except
        print(f"Could not parse:{line}")
````
