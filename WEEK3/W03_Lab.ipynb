### D6. A default value

1. Define `status_of(percent, warning_at=90)` that returns:

   - `"OK"` otherwise
2. Print `status_of(95)`, then `status_of(85)`, then `status_of(85, 80)`.

**Expected output:** `WARNING`, `OK`, `WARNING`

````python
# D6
"""The program defines a function status_of with 2 parameters and returns the status of the percentage"""
def status_of(percent, warning_at=90): #function with 2 parameters 
     if percent >= 100:
      return "OVER LIMIT" #status is overlimit when percent is greater than 100
     elif percent >= warning_at: 
      return "WARNING" #status is a Warning when percent is greater than warning 
     else:
      return "OK" #if above conditions not repected return Ok
print(status_of(95))
print(status_of(85))
print(status_of(85,80)) #printing the status by passing the parameters the value of warning_at is overwritten
````

````python
# D8
"""the program calculates the sum and display the key-value configuratiom path """
def total(*numbers): #calculates the sum of the nubers passed in
    return sum(numbers)

print(total(1,2,3)) #Assigned values for total
print(total(10,20,30,40))
print(total())
def configure(**settings):  #function setting 
 for key, value in settings.items(): #loops over the variable keyword abd prints tin key = value 
    print(f"{key}={value}")

configure(host= "10.0.0.1",port= 8080)  #prints host and post value 
````

1. Define `check_record(value, limit)` that works out `difference = value - limit` and

### C2. How many are over?

Write `count_over(*values, limit=100)` with a one-line docstring. It returns how many of the values
are over the limit.

| Call | Expected output |
|---|---|
| `count_over(87, 120, 45, 101)` | `2` |
| `count_over(87, 120, 45, 101, limit=50)` | `3` |

````python
# C2
def count_over(*values, limit=100): #defining function counr_over 
 """Returns how many of the given values are over the limit""" 
 return sum(1 for value in values if value > limit) #return sum of count_over if value is greater than limit 
print(count_over(87,120,45,101))
print (count_over(87, 120, 45, 101, limit=50))
print (count_over())
````
